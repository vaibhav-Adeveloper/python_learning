## Finding element in array.

```python
arrr
```

```python
np.where(arrr>5)
```

where() - Will return a tupple containing array as its element. - these array contains the index of the element which are greater than 5.

```python
type(np.where(arrr>5))
```

```python
np.count_nonzero(arrr)
```

count_nonzero(array) - will count all the non zero elements present in the array.

```python
np.nonzero(arrr)
```

you will not find 2, 2 as it is zero in array.

```python
arrr[1,2] = 0
```

```python
np.nonzero(arrr)
```

Now you also will not find 1, 2 as it is also now become zero.

```python
type(np.nonzero(arrr))
```

## Let see how Numpy takes less Space

```python
import sys
```

```python
py_arr = [0,4,55,2]
```

```python
np_arr = np.array(py_arr)
```

```python
sys.getsizeof(1) * len(py_arr)
```

normal python list takes 112 bytes for 4 elements.

```python
np_arr.itemsize * np_arr.size
```

numpy array takes 32 bytes for the same 4 elements

```python
np_arr.tolist()
```

tolist() - converts numpy array into python list.
