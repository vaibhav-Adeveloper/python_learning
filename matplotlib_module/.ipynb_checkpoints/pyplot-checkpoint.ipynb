# Bar Plot

##### Syntax for Bar plot

```python
x = ['C', 'C++', "Python", "Java"]
y = [60, 70, 85, 82]
```

```python
plt.title("Code World", fontsize=18)
plt.xlabel("Languages", fontsize=15)
plt.ylabel("Popularity %", fontsize=15)
plt.bar(x,y)
plt.show()
```

### Attributes of bar()

1. Width
2. color
3. align
4. edgecolor - to set boundary color of bars
5. linewidth - to set width of boundary of bars
6. linestyle - to set the style of boundary of bars
7. alpha - to set intensity of color of bars

```python
plt.bar(x,y, width=0.3, color="g")
plt.show()
```

Giving multiple colors

```python
c = ["b", "m", "y", "r"]
plt.bar(x,y, width = 0.3, color = c)
plt.show()
```

To move xlabel either on edge or center - align() - center or edge. 

```python
plt.bar(x,y, width=0.3, color=c, align="edge")
plt.show()
```

```python
plt.bar(x,y,width=0.3,color=c, edgecolor="green")
plt.show()
```

```python
plt.bar(x,y,width=0.3,color=c, edgecolor="green", linewidth = 3, linestyle = "dotted")
plt.show()
```

```python
plt.bar(x,y,width=0.3, color=c, edgecolor="green", alpha=0.6)
plt.show()
```

To set the label inside our bar graph - use label argument but also have to use legend() function for its reflection in graph

```python
plt.bar(x,y,width=0.3, color="m", edgecolor="green", label="Popularity")
plt.legend()
plt.show()
```

#### How To plot multiple bar graph in same bar graph

```python
plt.xlabel("Language", fontsize=15)
plt.ylabel("Popularity %", fontsize=15)
plt.title("Coding Stuff", fontsize=18)
x = ['C', 'C++', "Python", "Java"]
y = [60, 70, 85, 82]
z = [20,30,40,50]
width = 0.2
plt.bar(x, y, width, color="r")
plt.bar(x, z, width, color="y")
plt.show()
```

Now the both bar graph is overlaping each other  -  how to seperate them ?

We will create a array-p-consisting of index of x (language name)
1. p = [0, 1, 2, 3]
2. p1 = [0.2. 1.2, 2.2, 3.2]
3. use xticks(position, name, rotation) to rename x label - inspite of 0.0, 1.0, 2.0, 3.0 -- convert them into C, C++, Python, Java

```python
plt.xlabel("Language", fontsize=15)
plt.ylabel("Popularity %", fontsize=15)
plt.title("Coding Stuff", fontsize=18)

x = ['C', 'C++', "Python", "Java"]
y = [60, 70, 85, 82]
z = [20,30,40,50]

width = 0.2
p = np.arange(len(x))
p1 = [j+width for j in p]

plt.bar(p, y, width, color="r", label="Popularuty1")
plt.bar(p1, z, width, color="y", label="Popularuty2")

plt.xticks(p+0.1, x, rotation=25)
plt.xticks(p+width/2, x)
plt.legend()
plt.show()
```

Creating the horizontal bar graph

```python
plt.xlabel("Popularity %", fontsize=15)
plt.ylabel("Language", fontsize=15)
plt.title("Coding Stuff", fontsize=18)

x = ['C', 'C++', "Python", "Java"]
y = [60, 70, 85, 82]
z = [20,30,40,50]

width = 0.2
p = np.arange(len(x))
p1 = [j+width for j in p]

plt.barh(x, y, width, color="r", label="Popularuty1")
plt.barh(x, z, width, color="y", label="Popularuty2")

plt.legend()
plt.show()
```

# Scatter Plot

```python
day = [1,2,3,4,5,6,7]
no = [2,5,7,1,5,9,2]
plt.scatter(day, no, s=70)
plt.title("Scatter Plot", fontsize=18)
plt.xlabel("Day", fontsize=14)
plt.ylabel("No", fontsize=14)
plt.show()
```

Attributes of scatter()
1. color (color or c = ...)
2. size  (s = ...)
3. alpha (to set the transparency)
4. marker (to set the marker shape)
5. edgecolor (to set the boundary color)
6. linewidth (to set the boundary width)
7. cmap  (color map - to set the color of marker -- have range between 0 to 100)
8. colorbar() - show color bar range
9. to set the colorbar label - use --> set_label()

```python
day = [1,2,3,4,5,6,7]
no = [2,5,7,1,5,9,2]
col = ["r","g","y","m","b","g", "r"]
sizes = [90,120,150,180,200,220,290]
plt.scatter(day, no, color = col, s = sizes, alpha=0.4, marker="*", edgecolor="b", linewidth=2)
plt.title("Scatter Plot", fontsize=18)
plt.xlabel("Day", fontsize=14)
plt.ylabel("No", fontsize=14)
plt.show()
```

```python
day = [1,2,3,4,5,6,7]
no = [2,5,7,1,5,9,2]
colors = [30,10, 50, 90, 60, 45, 77]
sizes = [90,120,150,180,200,220,290]
plt.scatter(day, no, c = colors, s = sizes, cmap="viridis")
t = plt.colorbar()
t.set_label("Color Bar", fontsize=15)
plt.title("Scatter Plot", fontsize=18)
plt.xlabel("Day", fontsize=14)
plt.ylabel("No", fontsize=14)
plt.show()
```

```python
day = [1,2,3,4,5,6,7]
no = [2,5,7,1,5,9,2]
no2 = [3,4,6,9,3,7,4]
colors = [30,10, 50, 90, 60, 45, 77]
sizes = [90,120,150,180,200,220,290]
plt.scatter(day, no, c = colors, s = sizes, cmap="viridis")
plt.scatter(day, no2, c = "r", s = sizes, alpha=0.6)
t = plt.colorbar()
t.set_label("Color Bar", fontsize=15)
plt.title("Scatter Plot", fontsize=18)
plt.xlabel("Day", fontsize=14)
plt.ylabel("No", fontsize=14)
plt.show()
```

# Histogram Plot

#### Used when we have Large Amount of Data

```python
# creating the random data, create data from the ranfe 10 to 60 and data must be 50 in amount.
x = np.random.randint(10, 60, (50))
print(x)
```

```python
x = [20, 36, 43, 39, 51, 29, 44, 25, 38, 19, 13, 18, 57, 13, 55, 12, 28, 37, 44, 12, 58, 14, 29, 12,
 37, 24, 50, 24, 48, 57, 29, 31, 39, 18, 48, 11, 47, 45, 21, 59, 57, 21, 52, 24, 59, 52, 54, 38,
 54, 52]
```

```python
plt.hist(x, edgecolor='r')
plt.show()
```

```python
# about bins - means data ko Kin dono no kai beech mai dikhana hai.
# range - form with the help of tupple - means plot ki range kya hona chaiye (0,200)
l = [10, 20 ,30, 40, 50, 60]
plt.hist(x, bins=l, edgecolor='r')
plt.show()
```

```python
plt.hist(x, "auto", edgecolor='r')
plt.show()
```

```python
# setting range
plt.hist(x, "auto", (0,100), edgecolor='r')
plt.show()
```

```python
# cumulative parameter - reverse the frequency parameter
plt.hist(x, bins=l, edgecolor='r', cumulative=-1)
plt.show()
```

```python
# to change the bottom value of the y axis
plt.hist(x, edgecolor='r', bottom=15)
plt.show()
```

```python
# to change the align ment of the x axis values --> align= left,mid,right
plt.hist(x, bins=l, edgecolor='r', align="left")
plt.show()
```

```python
# to change the visual of the type of histogram -- use histtype
plt.hist(x, bins=l, edgecolor='r', histtype="step")
plt.show()
```

```python
# To change the orientation of the histogram
plt.hist(x, bins=l, edgecolor='r', orientation="horizontal")
plt.show()
```

```python
# to change the width of the bars in histogram - use rwidth
plt.hist(x, bins=l, edgecolor='r', rwidth=0.8)
plt.show()
```

```python
# to convert the histogram into log table
plt.hist(x, bins=l, edgecolor='r', log=True)
plt.show()
```

```python
# applying the label inside the graph with the help of the legend
plt.hist(x, bins=l, edgecolor='r', label = "Python")
plt.legend()
plt.show()
```

```python
# to draw the vertical line in the graph to seperate the data - use plt.axvline()
plt.hist(x, bins=l, edgecolor='r', label = "Python")
plt.axvline(45, color="y", label="line")
plt.legend()
plt.show()
```

```python
plt.hist(x, bins=l, edgecolor='r', label = "Python")
plt.axvline(45, color="y", label="line")
plt.legend()
plt.grid()
plt.show()
```
