About VIEW --> important concept.

```python
newdf2 = newdf
```

newdf2 is a view of newdf, not the copy of newdf

```python
newdf2[0][0] = 9898
```

```python
newdf
```

clearly seen that changes made in newdf2 get reflected in newdf.

1. newdf2 = newdf --> newdf pura ka pura copy hokai nya nhi banega in newdf2.
2. newdf2 is a view of newdf.
3. newdf2 ko as a pointer ki tarah samjh lo, jo ki newdf ki tarah point kar rha hai.
4. newdf2 change hoga toh newdf bhi change ho jaigaa.

now to copy the newdf to newdf2 such that both becomes independent to each other, changing one's value will not effect the
other value --> to do this we use copy() function.

```python
newdf2 = newdf.copy()
```

```python
newdf2[0][0] = 10000
```

this above warning is regarding to that view and copy concept --> warning is saying newdf2 is not a view of newdf, so on changing newdf2 it will not effect the newdf, instead changes made are now independent to each other.
